**Dataset Cleaning Function**

```python
TEXT_COLUMN = 'Comment'

def clean_bangla_text(Comment):

  Comment = str(Comment)

  #Removing URLs
  Comment = re.sub(r'http\S+|www\S+', '', Comment)

  #Removuing English words
  Comment = re.sub(r'[A-Za-z]+', ' ', Comment)

  #Removing English numbers
  Comment = re.sub(r'[0-9]+', ' ', Comment)

  #Eliminating Emojis
  Comment = emoji.replace_emoji(Comment, replace='')

  #Keeping only the Bangla characters
  Comment = re.sub(r'[^ঀ-৿\s।?!]', ' ', Comment)

  #Eliminating extra spaces
  Comment = re.sub(r'\s+', ' ', Comment).strip()

  return Comment


df['cleaned_text'] = df[TEXT_COLUMN].apply(clean_bangla_text)

#Eliminating empty rows
df = df[df['cleaned_text'].str.strip() != '']

#Removing rows with less than 5 characters
df = df[df['cleaned_text'].str.len() >= 5]

#Removing Duplicates
df.drop_duplicates(subset=['cleaned_text'], inplace=True)

#Index reset
df.reset_index(drop=True, inplace=True)

print(df[['cleaned_text']].head(30))
```

```python
output_file = '/content/cleaned_other_dataset.csv'

df.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f"\nCleaned dataset saved to: {output_file}")
```
